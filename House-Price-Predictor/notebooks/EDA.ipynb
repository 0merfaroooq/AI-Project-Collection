# 🏠 House Price Predictor — Exploratory Data Analysis

This notebook explores the Hyderabad residential house-price dataset used to train the prediction model.

**Sections:**
1. Load & inspect data
2. Data cleaning summary
3. Univariate analysis
4. Bivariate analysis
5. Correlation analysis
6. Locality-wise insights

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
sys.path.append('..')
from utils import clean_dataset

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)

df = pd.read_csv('../dataset/hyderabad_house_prices.csv')
df.head()
```

## 1. Basic Info

```python
print('Shape:', df.shape)
df.info()
```

```python
df.describe()
```

## 2. Data Cleaning Summary

```python
print('Missing values:\n', df.isnull().sum())
print('\nDuplicate rows:', df.duplicated().sum())

df_clean = clean_dataset(df)
print('\nRows before cleaning:', len(df))
print('Rows after cleaning:', len(df_clean))
```

## 3. Univariate Analysis

```python
plt.figure()
sns.histplot(df_clean['Price'] / 1e5, bins=40, kde=True, color='#2563eb')
plt.title('Price Distribution (Lakh INR)')
plt.xlabel('Price (Lakh INR)')
plt.show()
```

```python
plt.figure()
sns.countplot(data=df_clean, x='BHK', hue='BHK', palette='magma', legend=False)
plt.title('BHK Distribution')
plt.show()
```

## 4. Bivariate Analysis

```python
plt.figure()
sns.scatterplot(data=df_clean, x='Area_SqFt', y='Price', hue='BHK', palette='viridis', alpha=0.6)
plt.title('Price vs Area')
plt.show()
```

## 5. Correlation Analysis

```python
numeric_cols = ['Area_SqFt', 'BHK', 'Bathrooms', 'Parking', 'Property_Age', 'Floor_Number', 'Total_Floors', 'Price']
plt.figure(figsize=(9, 7))
sns.heatmap(df_clean[numeric_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

## 6. Locality-wise Insights

```python
loc_avg = df_clean.groupby('Location')['Price'].mean().sort_values() / 1e5
plt.figure(figsize=(10, 7))
loc_avg.plot(kind='barh', color='#059669')
plt.title('Location-wise Average Price (Lakhs)')
plt.xlabel('Average Price (Lakh INR)')
plt.show()
```

## Key Takeaways

- Premium localities (Jubilee Hills, Banjara Hills) command the highest average prices.
- Affordable localities (Uppal, LB Nagar) sit at the lower end of the price spectrum.
- Area (SqFt) has the strongest positive correlation with Price, as expected.
- The dataset required cleaning for a small number of missing values, duplicates, and outliers before modeling.
